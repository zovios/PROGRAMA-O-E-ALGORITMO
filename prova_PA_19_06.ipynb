```python
# QUESTÃO 02 – Soma dos Números (2 pontos)

# Solicite ao usuário um número inteiro positivo utilizando um laço de repetição, calcule e exiba a soma de todos os números de 1 até o número informado.

numero = int(input("digite um numero inteiro postivo: "))
soma = 0

while numero < 0:
  print("o numero tem que ser positivo!")
  numero = int(input("digite um numero inteiro: "))

for i in range(1, numero + 1):
  soma += i

  print(f"{soma}")

print(f"a soma dos numeros de 1 até {numero} é: {soma}")
```

```python
# QUESTÃO 03 – Tabuada (2 pontos)

# Solicite ao usuário um número inteiro e exiba sua tabuada de 1 até 10.

numero = int(input("digite um numero: "))

for i in range(1, 11):
  print(f"A tabuada de {numero} é igual a: {numero * i}")
```

```python
# QUESTÃO 04 – Fatorial (4 pontos)
# Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição.

numero = int(input("digite um número inteiro positivo: "))
fatorial = 1

if numero > 0:
  print("o número é positivo, pode prosseguir")
else:
  print("digite apenas números positivos")
  numero = int(input("digite novamente seu numero: "))

for i in range(1, numero + 1):
  fatorial *= i

  print(f"o fatorial de {numero} é: {fatorial}")
```

```python
# QUESTÃO 05 – Caixa Eletrônico (5 pontos)

#Um caixa eletrônico possui notas de:

#R$ 100
#R$ 50
#R$ 20
#R$ 10
#R$ 5
#R$ 2
#R$ 1

#Solicite ao usuário o valor do saque e informe a menor quantidade possível de notas para realizar o pagamento.

print("------CAIXA ELETRÔNICO-------")

valor = int(input("Digite o valor do saque: "))

if valor >= 100:
    notas_100 = 0
    while valor >= 100:
        notas_100 += 1
        valor -= 100
    print(f"{notas_100} nota(s) de R$100")

if valor >= 50:
    notas_50 = 0
    while valor >= 50:
        notas_50 += 1
        valor -= 50
    print(f"{notas_50} nota(s) de R$50")

if valor >= 20:
    notas_20 = 0
    while valor >= 20:
        notas_20 += 1
        valor -= 20
    print(f"{notas_20} nota(s) de R$20")

# R$ 10 notes
if valor >= 10:
    notas_10 = 0
    while valor >= 10:
        notas_10 += 1
        valor -= 10
    print(f"{notas_10} nota(s) de R$10")

if valor >= 5:
    notas_5 = 0
    while valor >= 5:
        notas_5 += 1
        valor -= 5
    print(f"{notas_5} nota(s) de R$5")

if valor >= 2:
    notas_2 = 0
    while valor >= 2:
       notas_2 += 1
       valor -= 2
    print(f"{notas_2} nota(s) de R$2")

if valor >= 1:
    notas_1 = 0
    while valor >= 1:
        notas_1 += 1
        valor -= 1
    print(f"{notas_1} nota(s) de R$1")
```
